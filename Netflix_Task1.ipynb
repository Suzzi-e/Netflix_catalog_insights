### Removing duplicate records and formatting inconsistencies

When `.info()` was first run, `string` was the datatype across all columns but one i.e., year. The string data type isn't exactly ideal when it comes to data analysis therefore we will format these inconsistencies for an easy analysis flow.

**LISTING INCONSISTENSIES**
1. **Incorrect datetime format** -
`date_added` column is currently a string data type, This hinders time-series analysis, filtering by month/year or sorting chronoliogically with strings.
2.  **Categorical columns as Strings** -
Columns like `type` and `rating` hold categorical data therefore converting them to `category` type is more memory-efficient and speeds up operations.

3. **Mixed String Columns** - Since the `duration` column contains both numeric lengths for movies (e.g., "90 min") and season counts for TV shows (e.g., "1 Season"), we can split this column into two new columns:
   * `duration_value`: A numeric column (integer) holding the count.
   * `duration_unit`: A categorical column containing the unit ('min' or 'Season').

```python
# Converting date_added column to datetime format
df_netflix['date_added'] = pd.to_datetime(df_netflix['date_added'])

#Coverting 'type' and 'rating' to categorical data types and saving the changes back to df_netflix
df_netflix = df_netflix.astype({'type': 'category','rating': 'category'})
```

```python
# Splitting 'duration' into value and unit
split_duration = df_netflix['duration'].str.split(' ', expand=True)

#Assigning new columns and cleaning up
df_netflix['duration_value'] = pd.to_numeric(split_duration[0], errors='coerce')
df_netflix['duration_unit'] = split_duration[1].str.replace('Seasons', 'Season').astype('category')

# Displaying the updated columns alongside the original columns
display(df_netflix[['title', 'type', 'duration', 'duration_value', 'duration_unit']].head(10))
```

```python
df_netflix.info()
```

### Standerdizing the Columns

Standerdizing columns like `country` , `rating`  and `type` ensures our dataset is clean, uniform and ready for data analysis.

```python
#Standerdizing the three columns
for col in ['country', 'rating', 'type']:
  df_netflix[col] = df_netflix[col].astype(str).str.strip()

#Ratings in Netflix are always in uppercase
#'country' and 'type' in title is neat as they are proper nouns
  if col == 'rating':
    df_netflix[col] = df_netflix[col].str.upper()
  else:
    df_netflix[col] = df_netflix[col].str.title()

#Converting to catgeory data types
#'rating' and 'country' had gone back to object during the standerdization.
df_netflix= df_netflix.astype({'country': 'category', 'rating': 'category', 'type':'category'})

#Displaying the columns after Standerdization
display(df_netflix[['title', 'country', 'rating']].head(10))
```
